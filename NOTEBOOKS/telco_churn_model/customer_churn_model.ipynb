```python
# Remove customer ID because it is only an identifier
X = X.drop("customerID", axis=1, errors="ignore")

# Find categorical columns
categorical_cols = X.select_dtypes(include=["object"]).columns.tolist()

print("Categorical columns:")
print(categorical_cols)
```

```python
X_encoded = pd.get_dummies(
    X,
    columns=categorical_cols,
    drop_first=True,
    dtype=int
)

print("Original X shape:", X.shape)
print("Encoded X shape:", X_encoded.shape)
```

```python
print("Remaining text columns:")
print(X_encoded.select_dtypes(include=["object"]).columns.tolist())
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_encoded,
    y,
    test_size=0.2,
    random_state=42,
    stratify=y
)

print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

```python
from sklearn.linear_model import LogisticRegression

logistic_model = LogisticRegression(
    max_iter=2000,
    random_state=42
)

logistic_model.fit(X_train, y_train)

print("Logistic Regression trained successfully!")
```

```python
y_pred = logistic_model.predict(X_test)
y_prob = logistic_model.predict_proba(X_test)[:, 1]

print("Predictions:", y_pred[:10])
```

```python
from sklearn.metrics import accuracy_score, classification_report

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
from sklearn.metrics import roc_auc_score

auc = roc_auc_score(y_test, y_prob)

print("ROC-AUC:", auc)
```

```python
from xgboost import XGBClassifier
```

```python
xgb_model = XGBClassifier(
    n_estimators=200,
    max_depth=4,
    learning_rate=0.05,
    random_state=42,
    eval_metric="logloss"
)
```

```python
xgb_model.fit(X_train, y_train)

print("XGBoost trained successfully!")
```

```python
xgb_pred = xgb_model.predict(X_test)
xgb_prob = xgb_model.predict_proba(X_test)[:, 1]

print("Predictions:", xgb_pred[:10])
```

```python
from sklearn.metrics import accuracy_score, classification_report, roc_auc_score

xgb_accuracy = accuracy_score(y_test, xgb_pred)
xgb_auc = roc_auc_score(y_test, xgb_prob)

print("XGBoost Accuracy:", xgb_accuracy)
print("XGBoost ROC-AUC:", xgb_auc)

print("\nClassification Report:")
print(classification_report(y_test, xgb_pred))
```

```python
print("MODEL COMPARISON")
print("----------------")
print("Logistic Regression Accuracy:", accuracy)
print("Logistic Regression ROC-AUC:", auc)
print("XGBoost Accuracy:", xgb_accuracy)
print("XGBoost ROC-AUC:", xgb_auc)
```

```python
results = pd.DataFrame({
    "Model": ["Logistic Regression", "XGBoost"],
    "Accuracy": [accuracy, xgb_accuracy],
    "ROC-AUC": [auc, xgb_auc]
})

results
```

```python
feature_importance = pd.DataFrame({
    "Feature": X_train.columns,
    "Importance": xgb_model.feature_importances_
})

feature_importance = feature_importance.sort_values(
    "Importance",
    ascending=False
)

feature_importance.head(15)
```

```python
import matplotlib.pyplot as plt

top_features = feature_importance.head(10)

plt.figure(figsize=(10, 6))
plt.barh(top_features["Feature"], top_features["Importance"])
plt.xlabel("Importance")
plt.ylabel("Feature")
plt.title("Top 10 Important Features")
plt.gca().invert_yaxis()
plt.show()
```
